# Count Files in Robot Image Folders

This notebook counts the number of image files in the `train`, `test`, and `val` folders for each robot type (cube, cylinder, etc.), then outputs the counts per robot type and overall totals.

## 1. Import Required Libraries
We will use `os` for directory operations.

```python
import os
from collections import defaultdict
```

## 2. Set Data Directory Path
Define the path to the main data directory containing robot type folders.

```python
# Set the main data directory path
DATA_DIR = os.path.abspath(os.path.join(os.getcwd()))
print(DATA_DIR)
```

## 3. List Robot Type Folders
List all robot type folders in the data directory.

```python
# List robot type folders (exclude non-directories and hidden folders)
original_types = [
    name for name in os.listdir(DATA_DIR)
    if os.path.isdir(os.path.join(DATA_DIR, name)) and not name.startswith('.')
    and name not in ['README_files', 'USMicroMagSet_For_tracking', 'augmented_data', '__pycache__', 'config']
]
augmentation_types = [
    f'augmented_data/{name}' for name in os.listdir(os.path.join(DATA_DIR, 'augmented_data'))
    if os.path.isdir(os.path.join(DATA_DIR, 'augmented_data', name)) and not name.startswith('.')
    and name not in ['README_files', 'USMicroMagSet_For_tracking', 'augmented_data', '__pycache__', 'config']
]
robot_types = sorted(original_types + augmentation_types)
print('Robot types:', robot_types)
```

## 4. Count Files in train, test, val Subfolders for Each Robot Type
Iterate through each robot type folder and count the number of files in its train, test, and val subfolders.

```python
# Count files in train, test, val for each robot type
counts = defaultdict(lambda: {'train': 0, 'test': 0, 'val': 0})
for robot in robot_types:
    for split in ['train', 'test', 'val']:
        img_dir = os.path.join(DATA_DIR, robot, 'images', split)
        if os.path.isdir(img_dir):
            num_files = len([
                f for f in os.listdir(img_dir)
                if os.path.isfile(os.path.join(img_dir, f))
            ])
            counts[robot][split] = num_files
```

## 5. Display Counts per Robot Type
Output the file counts for train, test, and val folders for each robot type in the specified format.

```python
# Display counts per robot type
for robot in robot_types:
    print(f"{robot}:")
    for split in ['train', 'test', 'val']:
        print(f"- {split}: {counts[robot][split]}")
    print()
```

## 6. Calculate and Display Overall Counts
Sum the counts across all robot types and display the overall train, test, and val file counts.

```python
# Calculate and display overall counts
overall = {'train': 0, 'test': 0, 'val': 0}
for robot in robot_types:
    for split in ['train', 'test', 'val']:
        overall[split] += counts[robot][split]
print("Overall:")
for split in ['train', 'test', 'val']:
    print(f"- {split}: {overall[split]}")
```
